## 2) Uso base di YOLO per rilevamento su una singola immagine

```python
from ultralytics import YOLO
import matplotlib.pyplot as plt

# Carichiamo un modello YOLO pre-addestrato (es. yolov8n)
model = YOLO("yolov8n.pt")

# Possiamo passare un percorso di file, una URL o un'immagine
results = model("https://ultralytics.com/images/bus.jpg")  # immagine di esempio

# results è una lista di oggetti Result; mostriamo il primo
res = results[0]

# Ultralytics fornisce direttamente .plot() che restituisce un'immagine annotata
annotated = res.plot()  # NumPy array (H,W,3) BGR

plt.figure(figsize=(8,8))
plt.imshow(annotated[..., ::-1])  # convertiamo BGR->RGB
plt.axis("off")
plt.title("YOLO detection")
plt.show()
```

```python
# Esempio (decommentare e mettere un path valido sul proprio sistema)
path_img = "../data/img/road.jpg"
results = model(path_img)
res = results[0]
annotated = res.plot()
plt.figure(figsize=(8,8))
plt.imshow(annotated[..., ::-1])
plt.axis("off")
plt.title("YOLO detection (immagine locale)")
plt.show()
```
